```python
# Análise de dados
import pandas as pd
import numpy as np

# Gráficos
import matplotlib.pyplot as plt


path = "https://raw.githubusercontent.com/squad-ada-lovelace/Desafio_07-Desafio-final/refs/heads/main/src/Sleep_health_and_lifestyle_dataset.csv"
df = pd.read_csv(path) 
```

```python
traducao_colunas = ['ID da pessoa', 'genero', 'idade', 'ocupacao', 'duracao do sono', 'qualidade do sono', 'tempo diario de atividade fisica', 'nivel de estresse',
                    'classificacao do IMC', 'pressao sanguinea', 'batimentos cardiacos','passos diarios', 'transtorno do sono']
colunas = df.columns

df = df.rename(columns=dict(zip(colunas, traducao_colunas)))

df
```

```python
#traduzindo possíveis respostas de gênero
traducao_genero ={
    'Male': 'masculino',
    'Female': 'feminino'}

df['genero'] = df['genero'].map(traducao_genero)

df
```

```python
#verificando possibilidades de ocupação
df['ocupacao'].unique()
```

```python
traducao_ocupacao = {
    'Nurse': 'enfermeiro(a)',
    'Doctor': 'medico(a)',
    'Engineer': 'engenheiro(a)',
    'Lawyer': 'advogado(a)',
    'Teacher': 'professor(a)',
    'Accountant': 'contador(a)',
    'Salesperson': 'vendedor(a)',
    'Scientist': 'cientista',
    'Software Engineer': 'engenheiro(a) de software',
    'Sales Representative': 'representante de vendas',
    'Manager': 'gerente'
}

df['ocupacao'] = df['ocupacao'].map(traducao_ocupacao)

df
```

```python
#traduzindo coluna transtorno do sono
traducao_transtorno_sono = {
    'Sleep Apnea': 'com transtorno',
    'Insomnia': 'com transtorno',
}

df['transtorno do sono'] = df['transtorno do sono'].map(traducao_transtorno_sono)

#atribuindo valor 'sem transtorno' aos registros com valores nulos da coluna 'transtorno do sono'
df['transtorno do sono'] = df['transtorno do sono'].fillna('sem transtorno')

df
```

```python
#criando nova coluna com duração do sono em minutos
df['duracao do sono (min)'] = df['duracao do sono'] * 60
df.drop(columns=['duracao do sono'], inplace=True)
```

```python
#divisão de pressao em duas variaveis
df[['pressao sistolica', 'pressao diastolica']] = df['pressao sanguinea'].str.split('/', expand=True)

#converter novas colunas pra variaveis numericas
df[['pressao sistolica', 'pressao diastolica']] = df[['pressao sistolica', 'pressao diastolica']].astype(int)

#excluir coluna original
df.drop(columns=['pressao sanguinea'], inplace=True)
df
```

```python
df.idade.max()
```

```python
intervalos = [0, 4, 6, 10]
rotulos = ['ruim', 'moderada', 'boa']
df['categoria qualidade do sono'] = pd.cut(df['qualidade do sono'], bins=intervalos, labels=rotulos)
df.head()
```

```python
#traduzindo classificação do IMC e unificando a categoria normal
traducao_imc = {
    'Overweight': 'sobrepeso',
    'Normal': 'peso normal',
    'Normal Weight': 'peso normal',
    'Obese': 'obesidade'
}

df['classificacao do IMC'] = df['classificacao do IMC'].map(traducao_imc)
df
```
